##Fancy Indexing
you can Pick elements from specific positions using a list/array of indices.

```python
import numpy as np
arr = np.array([10,20,30,40,50])

arr[[0,2,4]]
```

You can do this fancy indexing in 2D Matrix aswell

```python
matrix = np.arange(1,13).reshape(3,4)
print(matrix)
matrix[[0,1,2],[1,2,2]]
```

##Conditional Selection
Find values based on conditions or replace them.

```python
arr = np.array([5,12,18,7,30])
print(arr>10)
print(arr[arr>10])
print(np.where(arr>10))   #boolean indexing

print(np.where(arr>10,arr,0))   #put zero where value is false
```

##Sorting & Ordering
np.sort() → returns sorted array

np.argsort() → returns indices of sorted order

```python
arr = np.array([43,23,67,2,9,0])
print(np.sort(arr))
```

```python
np.argsort(arr)
```

##Concatenation & Stacking

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

np.concat([a,b])
```
