```python
# np.eye()       -> always 2D ##Can create rectangular matrices.
# np.identity()  -> always 2D ##Only square matrices.

# np.diag()      -> can return 1D or 2D
## If Input is 1D → Returns 2D Array
## If Input is 2D → Returns 1D Array

# np.diag() Gives: ValueError: Input must be 1- or 2-d.

# np.diagonal()   -> works with nD arrays #return 1d array
```

```python
import numpy as np
print(np.__version__)
```

```python
list1 = [1,2,3,4]
list2 = [5,6,7,8]
list1 + list2
```

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,7])
arr1 + arr2
```

```python
arr1 = np.array([1,2,3,4,True,"False"])
print(arr1)
# highest priority is str->float->int
```

```python
arr1 = np.array([[1,2,3,4]])
arr2 = np.array([[[4,5,6,7]]])
print(arr1.ndim)      #ndim is used to check number of dimensions
print(arr2.ndim)
```

```python
# Array Creation Methods
print(np.zeros(15))
```

```python
print(np.zeros([2,3]))    #row,columns
```

```python
print(np.ones(10))
```

```python
# for 2d 1 matrix
print(np.ones([2,3]))
```

```python
# full
print(np.full([5,5],9))
```

```python
# eye method makes diagonal element with 1 and rest fill 0
arr = np.eye(5)  #it makes square matrix if we gave only one value
print(arr)
print()

a2 = np.eye(2,4)   #rows,columns
print(a2)
```

```python
# identity matrix
# wht is difference btw identity method and eye method
iden_arr =np.identity(7)
print(iden_arr)
```

```python
# diag() is different.
# if input is 1D ->  Returns 2D array
arr = np.diag([1,2,3,4,5,6])
print(arr)
```

```python
# if input is 2D -> it returns diagnonal element from the ndim input
arr = np.diag([[1,2,3],
               [4,5,6],
               [7,8,9]])
print(arr)
```

```python
# if input is 3D then we have to use diagonal
arr = np.array(
    [[[1,2],[3,4]],
      [[4,5],[6,7]]]
)
np.diagonal(arr)
```

```python
# Basic Array attributes
arr = np.array([1,2,3,34,5,6],ndmin=2)  #ndim: minimum number of dimensions
print(arr)
print(arr.size)   #number of elements
print(arr.dtype)  #data type of array
print(arr.ndim)   #number of dimensions
print(arr.shape)  #size of each dimensions (row , column)
a_float = arr.astype(float) # convert integer into float or we can change astype(str)
print(a_float)
```
