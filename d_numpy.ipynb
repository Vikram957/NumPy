```python
# create a 2d array
arr = np.array([[1,2,3,4,5,6]])
print(arr)

# using flatten() - return a copy
# flatten() also converts a multidimensional array into a 1D array,
# but it always returns a copy.
flat_arr = arr.flatten()
flat_arr[1] = 1000
print(flat_arr)
print(arr)

# ravel() → Returns a view whenever possible.
# flatten() → Always returns a copy.

raveled_arr = arr.ravel()
raveled_arr[0] = 80000
print(raveled_arr)
print(arr)
```

```python
# CONCATENATE()
arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
arr3 = np.array([100,200,300,400])
value = np.concatenate([arr2,arr1,arr3])
print(value)
```

```python
arr1 = np.array([[1,2],[3,4]])
arr2 = np.array([[5,6],[7,8]])

arr = np.concatenate((arr1,arr2),axis = 1)  #axis=0 -> row , axis=1 ->column
print(arr)
```

```python
# Combine arrays vertically or horizontally   stacking
arr1 = np.array([[1,2],[3,4]])
arr2 = np.array([[5,6],[7,8]])

arr = np.stack((arr1,arr2),axis = 1)
print(arr)
```

```python
# VERTICAL STACK
a = np.array([[1,2,3],[1,4,5]])
b = np.array([[6,5,4],[2,4,3]])

print(np.vstack((a,b)))
```

```python
# HORIZONTAL STACK
a = np.array([[1,2,3],[1,4,5]])
b = np.array([[6,5,4],[2,4,3]])

print(np.hstack((a,b)))
```

```python
# FLIP ARRAY
arr = np.array([1,2,3,4])
arr1 = np.array([[1,2,3],[4,5,6]])
print(np.flip(arr))  #reverse
print(arr1)
print(np.flip(arr1))
```

```python
arr1 = np.array([1.4,2,3,4])
arr2 = np.array([2,3,4,5])
print(np.add(arr1,arr2))
print(np.subtract(arr1,arr2))
print(np.multiply(arr1,arr2))
print(np.divide(arr1,arr2))
print(np.power(arr1,arr2))
print(np.mod(arr1,arr2))     #gives reminder
print()

print(np.sqrt(arr1))
print(np.square(arr1))
print()

# trignometry
print(np.log10(arr1))
print(np.sin(arr1))
print(np.cos(arr1))
print(np.tan(arr1))
print()

print(np.floor(arr1))
print(np.ceil(arr1))
print(np.round(arr1))
```

```python
# Aggregate function
arr = np.array([1,2,3,4,5,6])
print(np.mean(arr))
print(np.median(arr))
print(np.average(arr))
print(np.std(arr))
print(np.var(arr))
print(np.min(arr))
print(np.max(arr))
print(np.argmin(arr))
print(np.argmax(arr))
print(np.sum(arr))
print(np.prod(arr))
```
