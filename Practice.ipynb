```python
import numpy as np
# Q1. Employee Salary Analysis – Array Creation & Slicing
# A company stores the monthly salaries of 10 employees in a NumPy array.
# import numpy as np

# salary = np.array([25000, 30000, 28000, 45000, 35000,
#                    40000, 32000, 50000, 27000, 38000])
# Display the salaries of the first 5 employees.
# Extract the salaries of the last 3 employees.
# Display salaries of employees at even index positions.
# Reverse the entire salary array without using a loop.
```

```python
salary = np.array([25000, 30000, 28000, 45000, 35000,
                   40000, 32000, 50000, 27000, 38000])
# Display the salaries of the first 5 employees.
print(salary[:5:])
```

```python
# Extract the salaries of the last 3 employees.
print(salary[-3::])
```

```python
# Display salaries of employees at even index positions.
print(salary[0::2])
```

```python
#  Reverse the entire salary array without using a loop.
print(salary[::-1])
```

```python
# Q2. Student Performance – Array Attributes
# A college stores students’ marks in 3 subjects using a NumPy array.
# marks = np.array([[85, 90, 78],
#                    [76, 88, 92],
#                    [90, 95, 89],
#                    [65, 70, 75]])
# Find the number of students and subjects.
# Check the total number of elements.
# Find the dimensions of the array.
# Calculate the average marks of all students.
# Reshape the array into a 2 × 6 array.
```

```python
marks = np.array([[85, 90, 78],
                   [76, 88, 92],
                   [90, 95, 89],
                   [65, 70, 75]])
print("number of students:",marks.shape[0])
print("number of marks:",marks.shape[1])
```

```python
# Check the total number of elements.
print(marks.size)
```

```python
# Find the dimensions of the array.
print(marks.ndim)
```

```python
 # Calculate the average marks of all students.
print(np.mean(marks))
```

```python
# Reshape the array into a 2 × 6 array.
print(marks.reshape(3,4))
```
