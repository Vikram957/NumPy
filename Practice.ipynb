```python
# Q4. Temperature Monitoring – Advanced Slicing
# A weather monitoring system records temperatures for 7 days at 4 different locations.
# temperature = np.array([
#     [30, 32, 31, 29],
#     [33, 35, 34, 30],
#     [29, 31, 32, 28],
#     [34, 36, 35, 32],
#     [31, 33, 30, 29],
#     [35, 37, 36, 33],
#     [32, 34, 33, 31]
# ])
# Extract temperatures recorded at all locations on the first 3 days.
# Extract the temperatures of the last 2 locations for all 7 days.
# Extract temperatures from day 2 to day 5 for location 2 and location 3.
# Extract alternate days from the entire array.
# Extract the last column without using a loop.

```

```python
temperature = np.array([
    [30, 32, 31, 29],
    [33, 35, 34, 30],
    [29, 31, 32, 28],
    [34, 36, 35, 32],
    [31, 33, 30, 29],
    [35, 37, 36, 33],
    [32, 34, 33, 31]
])

# Extract temperatures recorded at all locations on the first 3 days.
print(temperature[:3])
```

```python
#  Extract the temperatures of the last 2 locations for all 7 days.
print(temperature[:, 2:])    #(row,column)
```

```python
# Extract temperatures from day 2 to day 5 for location 2 and location 3.
print(temperature[1:5,1:3])
```

```python
# Extract alternate days from the entire array.
print(temperature[::2])
```

```python
# Extract the last column without using a loop.
print(temperature[:,-1:])
```

```python
# Q5. Inventory Management – Logical Array Creation & Attributes
# A warehouse tracks the stock quantity of 12 products.
# stock = np.array([50, 0, 25, 80, 10, 0, 45, 100, 15, 60, 0, 35])
# Find the shape, size, and number of dimensions.
# Create a new array containing 12 elements initialized with zeros.
# Extract products having zero stock using Boolean indexing.
# Extract every second product’s stock quantity.
# Reverse the array and identify the highest and lowest stock quantities

stock = np.array([50, 0, 25, 80, 10, 0, 45, 100, 15, 60, 0, 35])
# # Find the shape, size, and number of dimensions.
print("Shape:", stock.shape)
print("Size:", stock.size)
print("Dimensions:", stock.ndim)
```

```python
# Create a new array containing 12 elements initialized with zeros.
new_stock = np.zeros(12)
print(new_stock)
```

```python
# Extract products having zero stock using Boolean indexing.
print(stock[stock == 0])
```

```python
# Extract every second product’s stock quantity.
print(stock[::2])
```

```python
# # Reverse the array and identify the highest and lowest stock quantities
print(stock[::-1])      #reversed
print(np.max(stock))    #max
print(np.min(stock))     #min
```

Exercise 43:Find the intersection and union of two arrays

•	Intersection: elements present in both arrays.

•	Union: all unique elements from both arrays, sorted.


```python
a = np.array([1, 2, 3, 5, 7])
b = np.array([3, 4, 5, 6, 7])

inter = np.intersect1d(a,b)
print("intersection:",inter)

union = np.union1d(a,b)
print("union:",union)
```
